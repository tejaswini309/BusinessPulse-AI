```python
import pandas as pd
from pathlib import Path

# Paths
RAW_PATH = Path("../data/raw")
PROCESSED_PATH = Path("../data/processed")

print("Current Working Directory:")
print(Path.cwd())

print("\nRaw folder exists:", RAW_PATH.exists())
print("Processed folder exists:", PROCESSED_PATH.exists())
```

```python
customers = pd.read_csv(PROCESSED_PATH / "customers_clean.csv")
orders = pd.read_csv(PROCESSED_PATH / "orders_clean.csv")
order_items = pd.read_csv(PROCESSED_PATH / "order_items_clean.csv")
payments = pd.read_csv(PROCESSED_PATH / "payments_clean.csv")
products = pd.read_csv(PROCESSED_PATH / "products_clean.csv")
reviews = pd.read_csv(PROCESSED_PATH / "reviews_clean.csv")
sellers = pd.read_csv(PROCESSED_PATH / "sellers_clean.csv")
geolocation = pd.read_csv(PROCESSED_PATH / "geolocation_clean.csv")
category_translation = pd.read_csv(PROCESSED_PATH / "category_translation_clean.csv")

print("✅ All cleaned datasets loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments,
    "Products": products,
    "Reviews": reviews,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}

for name, df in datasets.items():
    print(f"{name:<22} Rows: {df.shape[0]:>8} | Columns: {df.shape[1]}")
```

```python
# Load original raw datasets
raw_customers = pd.read_csv(RAW_PATH / "olist_customers_dataset.csv")
raw_orders = pd.read_csv(RAW_PATH / "olist_orders_dataset.csv")
raw_order_items = pd.read_csv(RAW_PATH / "olist_order_items_dataset.csv")
raw_payments = pd.read_csv(RAW_PATH / "olist_order_payments_dataset.csv")
raw_products = pd.read_csv(RAW_PATH / "olist_products_dataset.csv")
raw_reviews = pd.read_csv(RAW_PATH / "olist_order_reviews_dataset.csv")
raw_sellers = pd.read_csv(RAW_PATH / "olist_sellers_dataset.csv")
raw_geolocation = pd.read_csv(RAW_PATH / "olist_geolocation_dataset.csv")
raw_category = pd.read_csv(RAW_PATH / "product_category_name_translation.csv")
```

```python
validation = pd.DataFrame({
    "Dataset": [
        "Customers",
        "Orders",
        "Order Items",
        "Payments",
        "Products",
        "Reviews",
        "Sellers",
        "Geolocation",
        "Category Translation"
    ],
    "Raw Rows": [
        len(raw_customers),
        len(raw_orders),
        len(raw_order_items),
        len(raw_payments),
        len(raw_products),
        len(raw_reviews),
        len(raw_sellers),
        len(raw_geolocation),
        len(raw_category)
    ],
    "Cleaned Rows": [
        len(customers),
        len(orders),
        len(order_items),
        len(payments),
        len(products),
        len(reviews),
        len(sellers),
        len(geolocation),
        len(category_translation)
    ]
})

validation["Status"] = validation.apply(
    lambda x: "✅ Match" if x["Raw Rows"] == x["Cleaned Rows"] else "⚠️ Changed",
    axis=1
)

validation
```

```python
# Check missing values in all cleaned datasets

for name, df in datasets.items():
    print(f"\n{name}")
    print("-" * 50)
    
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("✅ No missing values")
    else:
        print(missing)
```
