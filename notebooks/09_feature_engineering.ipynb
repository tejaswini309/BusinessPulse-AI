```python
import pandas as pd

customers = pd.read_csv("../data/processed/customers_clean.csv")
orders = pd.read_csv("../data/processed/orders_clean.csv")
order_items = pd.read_csv("../data/processed/order_items_clean.csv")
payments = pd.read_csv("../data/processed/payments_clean.csv")
products = pd.read_csv("../data/processed/products_clean.csv")
sellers = pd.read_csv("../data/processed/sellers_clean.csv")
reviews = pd.read_csv("../data/processed/reviews_clean.csv")
geolocation = pd.read_csv("../data/processed/geolocation_clean.csv")

print("Datasets Loaded Successfully")
```

```python
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"]
)

orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"]
)
```

```python
customer_orders = (
    orders
    .merge(customers, on="customer_id")
    .merge(payments, on="order_id")
)
```

```python
customer_orders.head()
```

```python
customer_spend = (
    customer_orders
    .groupby("customer_unique_id")["payment_value"]
    .sum()
    .reset_index(name="total_spend")
)

customer_spend.head()
```

```python
avg_order_value = (
    customer_orders
    .groupby("customer_unique_id")["payment_value"]
    .mean()
    .reset_index(name="avg_order_value")
)

avg_order_value.head()
```

```python
repeat_customer = (
    customer_orders
    .groupby("customer_unique_id")
    .size()
    .reset_index(name="order_count")
)

repeat_customer["repeat_customer"] = (
    repeat_customer["order_count"] > 1
).astype(int)

repeat_customer.head()
```

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days

orders[["delivery_days"]].head()
```

```python
orders["purchase_year"] = orders["order_purchase_timestamp"].dt.year
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.month
orders["purchase_day"] = orders["order_purchase_timestamp"].dt.day
orders["purchase_weekday"] = orders["order_purchase_timestamp"].dt.day_name()

orders.head()
```
